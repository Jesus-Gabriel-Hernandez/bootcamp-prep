```python
# Función para alterar cadenas de texto

# Tenemos un vector inicial, que por ejemplo quiero que sea todo mayúsculas
lista_nombres =["jesús","rosa","elena","antonio","marc","sara"]

# Defino una función que espera una lista
def list_to_uppercase(list):
# Creo una lista para poder guardar cada elemento ya alterado
    lista_nombres_mayus = []
# Recorro cada elemento, y guardo cada uno al nuevo vector
    for nombre in lista_nombres:
        lista_nombres_mayus.append(nombre.upper())
# Se lo imprimo por pantalla al usuario
    return lista_nombres_mayus
#Llamo o invoco la función dando como parámetro mi vector inicial
list_to_uppercase(lista_nombres)
    
```

```python
# Función para manejar números

# Tenemos un vector inicial, con números
lista_numeros =[1,2,3,4,5,5,7,8,9,19,32,5346,856,0]

# Defino una función que espera una lista
def list_greater_five(list):
# Creo una lista para poder guardar cada elemento mayor a 5
    lista_mayor_cinco = []
# Recorro cada elemento, según un condicional
    for numero in lista_numeros:
# Si es mayor a cinco lo guardo en el nuevo vector
        if(numero >5):
            lista_mayor_cinco.append(numero)
# Se lo imprimo por pantalla al usuario
    return lista_mayor_cinco
#Llamo o invoco la función dando como parámetro mi vector inicial
list_greater_five(lista_numeros)
    
```
