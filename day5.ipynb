```python
# Día 5 Exploración de datos con estadística + visualización

# Este día será solamente una expansión del día anterior, por lo que, usaremos los mismos casos

# Usaremos el método boxplot
import pandas as pd
import matplotlib.pyplot as plt

# Creamos el, conjunto de datos
conjunto_datos = {
    "Producto": ["Fresas","Leche","Huevos","Huevos","Leche","Fresas"],
    "Ventas": [60,225,245,52,143,15],
    "Ciudad": ["Barcelona","Barcelona","Galicia","Galicia","Barcelona","Sevilla"]
}

# Creamos una variable para crear nuestro marco de datos, o DataFrame con el diccionario de antes
df = pd.DataFrame(conjunto_datos)


# El .describe proporciona datos como:  media; desviación estándar, valores mínimos y máximos, percentiles de las categorías numéricas.
df.describe() # Yo solo lo pongo porque nunca viene mal, hacer una pequeña lectura de los datos, para ver, como sería la gráfica

# Ahora, haremos unos gráficos, para datos númericos, y otros categóricos

# Importante, boxplot solo interpreta datos tipo número o número + categoría, pero no solo categoría
plt.boxplot(df["Ventas"])

plt.title("Boxplot empleando matplotlib")
plt.ylabel("Unidades vendidas")
plt.show()


# Haremos un ejemplo con ambas categorías

# column solo permite datos numéricos, by es similar a SQL cuando hacemos un GROUP BY para ordenar,
#en nuestro caso, haremos un gráfico de las ventas por cada cuidad

df.boxplot(column="Ventas", by="Ciudad")

# .suptitle("") elimina el título automático que pone pandas, uno estilo <xlabel ="", y datos>
plt.suptitle("") 
plt.ylabel("Unidades vendidas") 
plt.show()
```

# Como podemos ver, matplotlib forma el gráfico de una forma distinta

# Eso se debe a lo siguiente por lo que he podido ver por Internet

Matplotlib dibuja el boxplot de esquemática. 

Lo que realmente muestra es:

Caja (box): va del percentil 25 (Q1) al percentil 75 (Q3), 

    sería decir dónde está la mayoría de datos, si ordenaramos (menos a más) dichos datos, la caja cubre el 25% al 75%, ni muy bajo, ni muy bajo
Línea dentro de la caja: es la mediana (50%),

    o la mitad que determina si el dato está por encima o no.
Bigotes (whiskers): Sería como decir desde el valor mínimo, hasta el máximo, marcan límites, todo lo de ese rango es un dato normal

Puntos fuera de los bigotes: serían outliers (valores atípicos), serían datos extraños, casos especiales, o sin patrón.

Como el parámetro df["Ventas"] contiene: [60, 225, 245, 52, 143, 15] y da lo siguientes datos:

    Mediana: 101.5
    Caja: entre ≈ 52 y ≈ 225
    Bigotes: alcanzan 15 y 245

    Outliers: no hay, porque todos los valores entran en el rango esperado, 
    si hubiera 1000, estaría por no estar en el rango mencionado.

# ¿Por qué parece extraño?
Como son pocos datos y algunos valores son bastante altos, la caja queda muy pequeña y los bigotes dominan la figura. Por eso se ve más como un “diagrama eléctrico” que una caja. 
# En seaborn, en cambio, veremos un gráfico más familiar.

```python
# Ahora veremos un ejemplo de boxplot con seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ahora crearemos uno
df_propio = pd.DataFrame({
    "Ventas": [154,354,357,235,677],
    "Ciudad": ["Madrid","Galicia","Aragón","Galicia","Madrid"]
})

# Gráfico con datos numéricos 
sns.boxplot(y="Ventas", data=df_propio)
plt.title("Ventas totales")
plt.xlabel("Ventas")
plt.show()


# Gráfico con datos numéricos + categoría
# En este caso, nos piden, categoría, dato numérico, y el marco de datos
sns.boxplot(x="Ciudad",y="Ventas",data=df_propio)

# Todo lo demás se puede con el matplotlib
plt.title("Ventas totales")

plt.ylabel("Ventas")
plt.xlabel("Frecuencia")
plt.show()


```
