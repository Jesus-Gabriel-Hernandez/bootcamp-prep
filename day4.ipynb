```python
#Día 4

# Ejercicio básico de matrices

# Creo una lista con notas
lista_notas = [
    [8,5,3,6,9],
    [1,6,10,5,7],
    [6,3,6,8,1],
    [7,4,8,8,4],
    [1,6,8,3,5],
    [6,7,4,5,5]
]

# Tras eso creo otra de alumnos
lista_alumnos = ["Ana","César","María","Pedro","Sandra","Diego"]

# En este caso, no es necesario hacer un doble bucle

# Simplemente con la función in range y la longitud de cada lista nos permite ir por cada índice
# La función min es para evitar si por ejemplo no hubieran las notas de un alumno o viceversa

for x in range(min(len(lista_alumnos), len(lista_notas))):
# Para que itere por cada elemento de las listas agarramos la varible x
    alumno = lista_alumnos[x]
    notas = lista_notas[x]

# Esto permite saber al usuario las notas de cada alumno, promedio,máxima nota y mínima
    print("El alumno", alumno, "tiene estas notas ",notas)

    maxima_nota = max(notas)
    minima_nota = min(notas)

    print("Su máxima nota es:",maxima_nota,"la mínima es:",minima_nota)
    promedio = sum(notas)/ len(notas)
    print("El promedio de las notas del alumno:",alumno,"es:", promedio,"\n")

```

```python
# Ejercicio de bucles y funciones

def es_par(numero):
    # Primero, comprobamos que sea un número válido
    if not numero.isnumeric():
        print("Solo se aceptan números")
        return  # Salimos de la función si no es válido

    # Convertimos a entero
    numero_int = int(numero)

    if numero_int % 2 == 0:
        print("El número", numero_int, "es par\n")
    else:
        print("El número", numero_int, "es impar\n")

    # Ahora mostramos todos los números del 1 al 20, solo para experimentar bucles dentro de una función
    print("Comprobación del 1 al 20:")
    for x in range(1, 21):
        if x % 2 == 0:
            print("El número", x, "es par")
        else:
            print("El número", x, "es impar")


# Llamamos a la función con lo que ingrese el usuario
respuesta_usuario = input("Ingresa un número: ")
es_par(respuesta_usuario)
```

```python
# Parte de análitica (introducción usando Python)

# Primero importamos la libreria de Pandas, yo usaré el as pd, para evitar escribir pandas.operacion1, pandas.operacion...

import pandas as pd

# Para luego hacer un gráfico usaremos o el matplotlib o el seaborn, para este caso haré el matplotlib
import matplotlib.pyplot as plt

# Creamos el, conjunto de datos
# Que al fin y al cabo es como si estuvieramos usando diccionarios

conjunto_datos = {
    "Producto": ["Fresas","Leche","Huevos","Huevos","Leche","Fresas"],
    "Ventas": [60,225,245,52,143,15],
    "Ciudad": ["Barcelona","Barcelona","Galicia","Galicia","Barcelona","Sevilla"]
}

# Creamos una variable para crear nuestro marco de datos, o DataFrame con el diccionario de antes
df = pd.DataFrame(conjunto_datos)

# El .head() te muestra tu marco de datos similar a una tabla sql, que te indica las series, (columnas y filas)
df.head()

# Ahora, haremos unos gráficos, para datos númericos, y otros categóricos

# Si son números usamos el .hist que usa parámetros de: marco de datos o columnas de este, el nivel de detalle
# opcionales el color y transparencia

plt.hist(df["Ventas"], bins=5, color="green", alpha=0.5)

# .xlabel() es para definir la leyenda del eje x, en este caso las ventas
plt.xlabel("Ventas")
# .ylabel() es para definir la layenda del eje y, en este caso la frecuencia
plt.ylabel("Frecuencia")
# .title() es para poner un título descriptivo al gráfico
plt.title("Ventas totales")
# .show() siempre es necesario para mostrar el gráfico realizado
plt.show()

# Ahora haremos un gráfico de las ventas por cuidad

# Lo único que destacamos es que no usamos el .hist, por no ser números, y que usamos plot definiendo el tipo del gráfico
ventas_ciudad = df.groupby("Ciudad")["Ventas"].sum()

ventas_ciudad.plot(kind="bar", color="orange", alpha=0.8)
plt.ylabel("Ventas")
plt.title("Ventas por ciudad")
plt.show()

# Pero usualmente es, ¿usas números? Usa .hist, si no usas el .plot(kind="bar") o el .plot()

```

```python
# Ahora probaremos el seaborn
import pandas as pd
import matplotlib.pyplot as plt

# Lo único para comentar es que si queremos usar seaborn, se ha de importar también el matplotlib
import seaborn as sns

# pd.read_csv() para leer archivos csv, se pueden tantos externos, o como uno del propio equipo,
# en mi caso solo dejo el cómo, pero no se usará para este caso
dataframe_externo = pd.read_csv("https://raw.githubusercontent.com/andresem/Datos/main/DatosAcciones.csv")

# Ahora crearemos uno

df_propio = pd.DataFrame({
    "Ventas": [154,354,357,235,677],
    "Ciudad": ["Madrid","Galicia","Aragón","Galicia","Madrid"]
})

# Gráfico con datos numéricos

# La línea es muy parecida al ejemplo del matplotlib: plt.hist(df["Ventas"], bins=5, color="green", alpha=0.5)
sns.histplot(df_propio["Ventas"],bins=20, color="orange")

# Todo lo demás se puede con el matplotlib
plt.title("Ventas totales")
plt.xlabel("Ventas")
plt.ylabel("Frecuencia")
plt.show()


# Ahora haremos con datos categóricos

# Aquí usamos el countplot, y los argumentos son: columna y el marco de datos
sns.countplot(x="Ciudad",data=df_propio)
# Todo lo demás se puede con el matplotlib

plt.title("Ventas por ciudad")
plt.xlabel("Ciudad")
plt.ylabel("Frecuencia")
plt.show()
# Similar al matplotlib: datos numéricos= histplot,
# datos categóricos = countplot, ambos mostrando distribución =boxplot 

```
