# Investigando sobre gráficos y tipos, encontré con el de tipo violín
    El gráfico violín combina la información resumida de un boxplot con la forma de la distribución de los datos,
    permitiendo ver no solo la mediana y los cuartiles, sino también dónde se concentran o dispersan las observaciones.
# Por lo que lo implementaremos

```python
# Día 6 descripción de estadística + visualizaciones

# En este día nos enfocaremos en el aspecto de analizar los datos, o la análitica descriptiva
# Operaciones como la media, desviación estándar, moda, junto otros ejemplos

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Creamos el marco de datos (los datos se han cambiado para tener una mejor visión de los mismos)
conjunto_datos = {
    "Ventas": [60,40,80,100,80,70],
    "Ciudad": ["Barcelona","Barcelona","Galicia","Galicia","Barcelona","Sevilla"]
}

# Creamos una variable para crear nuestro marco de datos, o DataFrame con el diccionario de antes
df = pd.DataFrame(conjunto_datos)

# Previamente, habíamos visto la función .describe(), que nos dan elementos como la media o la desviación
# Pero ahora compararemos el .describe() y las operaciones antes mencionadas

media_datos = df["Ventas"].mean()
mediana_datos = df["Ventas"].median()
moda_datos = df["Ventas"].mode()[0]
desviacion_datos = df["Ventas"].std()
varianza_datos = df["Ventas"].var()

print("La media de las ventas es: ", media_datos)

print("La mediana de las ventas es: ", mediana_datos)

print("El valor más repetido (moda) de las ventas es: ", moda_datos)

print("La desviación estándar de las ventas es: ", desviacion_datos)

print("La variación de datos de las ventas es: ", varianza_datos)

print("\n")
df.describe()

```

# Hagamos una pequeña comparación

    Con el describe las operaciones que nos proporciona aparte de las que nosotros declaramos, son:
    
    count, que nos cuenta cuántos elementos son en total, el valor mínimo y máximo, y los percentiles del 25%,50% y 75%, el valor por debajo del 25& de      los datos, el valor que se encuentra entre los datos bajos y altos,y el valor que está por denajo del 75% de los datos, por lo que el describe es        una gran opción para hacer un análisis de los datos
# Veamos el gráfico de estos datos, a ver que nos pueden mostrar

```python

# Día 6 descripción de estadística + visualizaciones

# En este día nos enfocaremos en el aspecto de analizar los datos, o la análitica descriptiva
# Operaciones como la media, desviación estándar, moda, junto otros ejemplos

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Creamos el marco de datos (los datos se han cambiado para tener una mejor visión de los mismos)
conjunto_datos = {
    "Ventas": [60,40,80,100,80,70],
    "Ciudad": ["Barcelona","Barcelona","Galicia","Galicia","Barcelona","Sevilla"]
}

# Creamos una variable para crear nuestro marco de datos, o DataFrame con el diccionario de antes
df = pd.DataFrame(conjunto_datos)

# Previamente, habíamos visto la función .describe(), que nos dan elementos como la media o la desviación
# Pero ahora compararemos el .describe() y las operaciones antes mencionadas

media_datos = df["Ventas"].mean()
mediana_datos = df["Ventas"].median()
moda_datos = df["Ventas"].mode()[0]
desviacion_datos = df["Ventas"].std()
varianza_datos = df["Ventas"].var()

plt.hist(df["Ventas"],bins=20)
plt.title("Historigrama")
plt.xlabel("Unidades vendidas")
plt.ylabel("Frecuencia")
plt.show()

# Gráfico violín con seaborn

# Podemos darnos cuenta que la estructura es muy similar a la de boxplot, ya que nos piden la columna y nuestro marco de datos
sns.violinplot(y="Ventas",data=df)
plt.title("Gráfico violín")
plt.show()
```

# En el gráfico 1 (histograma)
    La mayoría de las ventas se concentran entre 60 y 100 unidades.
    Se puede observar un “salto” alrededor de 80, que aparece con mayor frecuencia.
    Podemos apreciar que es una distribución NO uniforme, debido a que hay datos con mayor aparición y otros con una en menor cantidad
    
# Gráfico de violín
    La parte más ancha indica dónde se concentran más datos, mostrando la densidad de las ventas.
    La línea blanca marca la mediana
    En este caso, varios datos se encuentran entre el rango de 75 y 80 en la parte superior
