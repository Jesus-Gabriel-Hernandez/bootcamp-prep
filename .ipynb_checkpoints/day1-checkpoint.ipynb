```python
# Calcular media números
# Creo un vector y una variable para poder sumar cada elemento 
lista_num = [1,2,3,4,5]
add_elements = 0

# Si la lista no tiene 5 números avisa al usuario
if(len(lista_num) !=5):
    print("Lista mínimo con 5 números")
else:
# Si no, sumo los elementos y divido por la longitud del vector, devolviendo el resultado
    for elements in lista_num:
        add_elements += elements
        result = add_elements / len(lista_num)
print(result)
```
