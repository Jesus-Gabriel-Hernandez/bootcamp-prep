```python
# Día 7 Repaso de los elementos dados

# En este día solamente vamos a aplicar lo que se ha podido observar en la última parte de análitica
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

# Para este caso el marco de datos, será de 4 columnas
conjunto_datos = {
    "Producto": ["Lechuga","Naranja","Yogur","Queso","Zanahoria","Sandía"],
    "Ventas": [120,130,133,121,122,139],
    "Precio": [0.92,1.20,1,8.60,1.15,1.45],
    "Categoría" : ["Verdura","Fruta","Lácteo","Lácteo","Verdura","Fruta"]
}

# Lo pasamos como un marco de datos
df_productos = pd.DataFrame(conjunto_datos)

# En este caso no es necesario, pero siempre viene bien el revisar la estructura del marco de datos

#df_productos.info()
#df_productos.head()
#df_productos.describe()

# Histograma

# Haremos un simple histograma para ver la frecuencia de las ventas
plt.hist(df_productos["Ventas"],bins=10,label="Unidades vendidas del producto")
plt.ylabel("Frecuencia")
plt.xlabel("Unidades vendidas")
plt.legend()
plt.show()

# Boxplot
# Aquí será un gráfico con el mismo objetivo, solo que más visual para percibir mejor la tendencia
sns.boxplot(y=df_productos["Ventas"])
plt.ylabel("Frecuencia")
plt.xlabel("Unidades vendidas")
plt.show()

# Gráfico para comparar elementos numéricos con categóricos
sns.boxplot(x="Categoría", y="Ventas", data=df_productos)
plt.title("Ventas por categoría")
plt.show()

# Gráfico de violín

# Para notar la proporción de datos similares, lo mejor sería un violín, de este modo podremos ver el rango más común en los precios
sns.violinplot(y="Precio",data=df_productos)
plt.title("Precios de los productos")
plt.show()
```
