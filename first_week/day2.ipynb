```python

# Día 2

# Creo una lista frutas y la muestro
fruits_list = []

print("Lista original ",fruits_list)
# Agrego frutas como Sandía, Manzana y Banana

fruits_list.append("Sandía")
fruits_list.append("Manzana")
fruits_list.append("Banana")
fruits_list.append("Arándanos")
print("Lista tras el append", fruits_list)

# Elimino un elemento
fruits_list.remove("Manzana")
print("Lista tras el remove", fruits_list)

# Ordeno dicha lista
fruits_list.sort()
print("Lista ordenada", fruits_list)

# Prueba del slicing, en este caso muestro desde el primer elemento hasta el segundo
print("probando el slicing ",fruits_list[1:3])
```

```python
# Preguntar edad

# Pregunto la edad al usuario
user_age= input("Escribe un número y te doy un mensaje: ")

# Si no es tipo númerico. comento que escriba un número
if(not user_age.isnumeric()):
    print("ingresa un número")
    user_age= input("Escribe un número y te doy un mensaje: ")
else:
# Tras eso, paso el número a tipo int, y muestro mensajes según rangos
    int_age = int(user_age) 
    if(int_age < 18):
        print("Menor de edad", int_age)
    elif(int_age >= 18 and int_age <= 65):
        print("Adulto",int_age)
    elif(int_age > 65):
        print("Senior",int_age)
    else:
        print("Edad ilógica o fallo del sistema")
```

```python
# Reto nota alumnos

# Creo un vector de notas
vector_notas = [7, 8, 9, 5, 10]

# Cómo sacar la mayor nota
mayor_nota = max(vector_notas)
print(mayor_nota)

# Cómo sacar la mínima nota
menor_nota = min(vector_notas)
print(menor_nota)

# Cómo sacar el promedio
promedio_notas = (sum(vector_notas))/len(vector_notas)
print(promedio_notas)
```

```python

# Reto 2, bucle + condicionales

# Creo un vector y lo muestro
frutitas =["Manzana","Pera","Naranja"]
print(frutitas)

# Pregunto al usuario si quiere añadir, eliminar algún elemento al vector, o si solo lo quiere mostrar
user_option = input("¿Qué quieres hacer? Opciones: Añadir, eliminar o mostrar")

# Paso a minúsculas las opciones para manejar mejor condicionales

real_option = user_option.lower()
opciones_validas =["añadir","eliminar","mostrar"]
# Un bucle para que se repita si el usuario escribió alguna opción válida

while real_option in opciones_validas:
    if(real_option=="añadir"):
        fruta= input("Di el elemento a añadir: ")
        frutitas.append(fruta)
    elif(real_option=="eliminar"):
        fruta= input("Di el elemento a destruir: ")
        if(fruta in frutitas):
            frutitas.remove(fruta)
        else:
            print("La fruta no existe")
    elif(real_option=="mostrar"):
        print(frutitas)
#Pregunto nuevamente para que el programa termine
user_option = input("¿Qué quieres hacer? Opciones: Añadir, eliminar,mostrar u otra cosa para salir:")
print("Programa terminado")
```
